```python
# ============================================================
# Case 1 Notebook: Distances (GC vs Sea-only) — Rotterdam→Yokohama
# Reproducible pipeline producing tables + figures
# ============================================================

# --- 0. Imports & environment info ---
import os, math, json, textwrap, warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# Great-circle distance helper (no internet needed)
try:
    from geopy.distance import geodesic
    def gc_nm(a, b):
        # a,b = (lat, lon)
        return geodesic(a, b).nautical
except Exception:
    # fallback: haversine in nautical miles
    R_km = 6371.0088
    KM2NM = 0.539956803
    def gc_nm(a, b):
        lat1, lon1 = map(math.radians, a)
        lat2, lon2 = map(math.radians, b)
        dlat, dlon = lat2-lat1, lon2-lon1
        h = math.sin(dlat/2)**2 + math.cos(lat1)*math.cos(lat2)*math.sin(dlon/2)**2
        return 2*R_km*math.asin(math.sqrt(h))*KM2NM


# Searoute import (used for sea-only distances)
try:
    import searoute
    HAVE_SEAROUTE = True
except Exception as e:
    HAVE_SEAROUTE = False
    print("WARNING: searoute not installed. Sea-only paths will fail.")
    print("Install with: pip install searoute")
    print("Error:", e)
```

```python
# --- 1. I/O configuration ---
OUT_TABLES = Path("tables")
OUT_FIGS   = Path("figures")
OUT_LOGS   = Path("logs")
for d in [OUT_TABLES, OUT_FIGS, OUT_LOGS]:
    d.mkdir(parents=True, exist_ok=True)

WARNLOG = []

def log_warn(msg):
    WARNLOG.append(msg)
    print("WARN:", msg)
```

```python
# --- 2. Utilities: longitude normalization & dateline guards ---

def norm_lon(lon):
    """Normalize lon to [-180, 180)."""
    lon = ((lon + 180) % 360) - 180
    return lon

def crosses_dateline(lon1, lon2):
    """True if segment crosses antimeridian in normalized space."""
    lon1n, lon2n = norm_lon(lon1), norm_lon(lon2)
    return abs(lon2n - lon1n) > 180

def insert_dateline_guards(points, guard_lon=179.6):
    """
    Inserts micro-waypoints when a leg crosses the dateline.
    Strategy: add (lat, +guard_lon) then (lat, -guard_lon) to split.
    """
    out = [points[0]]
    for (lat1, lon1), (lat2, lon2) in zip(points[:-1], points[1:]):
        if crosses_dateline(lon1, lon2):
            # linear interpolation latitude at crossing (simple & stable here)
            mid_lat = 0.5 * (lat1 + lat2)
            out.append((mid_lat,  guard_lon))
            out.append((mid_lat, -guard_lon))
        out.append((lat2, lon2))
    return out

def polyline_nm(poly_latlon):
    """Sum GC distance along a polyline (lat,lon)."""
    if len(poly_latlon) < 2:
        return 0.0
    return sum(gc_nm(a, b) for a, b in zip(poly_latlon[:-1], poly_latlon[1:]))
```

```python
# --- 3. Searoute wrapper + suspicious-leg guard ---

def searoute_leg_latlon(a, b):
    """
    Returns sea-only polyline for leg a->b as list of (lat,lon).
    Handles common searoute API variants.
    """
    if not HAVE_SEAROUTE:
        raise RuntimeError("searoute not available")

    # searoute expects (lon,lat) for start/end in many versions
    start = (a[1], a[0])
    end   = (b[1], b[0])

    # try a few known patterns
    try:
        res = searoute.searoute(start, end)
    except TypeError:
        res = searoute.searoute(start, end, units="naut")

    # res can be dict-like GeoJSON
    coords = None
    if isinstance(res, dict):
        try:
            coords = res["features"][0]["geometry"]["coordinates"]
        except Exception:
            coords = res.get("geometry", {}).get("coordinates", None)
    else:
        # sometimes it's a GeoDataFrame/Feature-like object
        try:
            coords = res.geometry.iloc[0].coords[:]
        except Exception:
            coords = None

    if coords is None:
        raise RuntimeError("Unable to extract searoute coordinates; update extractor.")

    # coords are (lon,lat) -> convert
    poly = [(c[1], norm_lon(c[0])) for c in coords]
    return poly

def sea_leg_with_guard(a, b, ratio_thresh=3.0):
    """
    Compute sea-only leg; if searoute returns something insane
    (polylen >> GC), fallback to GC for that leg.
    """
    gc_len = gc_nm(a, b)
    poly = searoute_leg_latlon(a, b)
    poly_len = polyline_nm(poly)

    if (poly_len > ratio_thresh * gc_len) or (len(poly) < 3):
        log_warn(
            f"suspicious searoute leg {a}->{b}: poly={poly_len:.1f} nm, "
            f"gc={gc_len:.1f} nm; using GC."
        )
        return [a, b], gc_len
    return poly, poly_len
```

```python
# --- 4. Scenario definition (macro-waypoints) ---
# Coordinates are (lat, lon)

PHILOSOPHY_DESC = {
    "service": "Service-guided: approximates typical liner/bulk trunk logic via major hubs & chokepoints.",
    "bluewater": "Bluewater-oriented: favors longer open-ocean legs and fewer coastal/strait constraints where feasible.",
    "coast": "Coast/channel-guided: adheres more closely to constrained lanes, straits, and coastal guidance."
}

# Baseline OD
ROTTERDAM = (51.95, 4.14)
YOKOHAMA  = (35.45, 139.65)

WAYPOINTS = {
    "SUEZ": {
        # (a) Service-guided: classic trunk via English Channel, Gibraltar, Suez, Bab-el-Mandeb,
        # Sri Lanka gateway, Malacca/Singapore, SCS, Japan approach.
        "VAR1_service": [
            ROTTERDAM,
            (50.5, 1.0),          # English Channel exit
            (36.0, -5.6),         # Gibraltar
            (30.0, 32.6),         # Suez Canal
            (12.6, 43.3),         # Bab-el-Mandeb
            (6.0, 80.6),          # Sri Lanka gateway (Colombo lat/lon)
            (1.2, 103.5),         # Malacca/Singapore
            (15.0, 120.0),        # South China Sea trunk
            (35.0, 140.0),        # Tokyo approach gate
            YOKOHAMA
        ],

        # (b) Bluewater: still must use Suez + Bab-el-Mandeb + Malacca,
        # but uses more open Indian Ocean + more direct Pacific approach.
        "VAR2_bluewater": [
            ROTTERDAM,
            (48.0, -10.0),        # open Atlantic leg
            (36.0, -5.6),         # Gibraltar (still mandatory)
            (30.0, 32.6),         # Suez Canal
            (12.6, 43.3),         # Bab-el-Mandeb
            (0.0, 70.0),          # open Indian Ocean pivot (south of Sri Lanka)
            (1.2, 103.5),         # Malacca
            (20.0, 130.0),        # open W-Pacific pivot
            (35.0, 140.0),        # Japan gate
            YOKOHAMA
        ],

        # (c) Coast/channel-guided: more constrained coastal progression in Med/Red Sea/Arabian Sea.
        "VAR3_coast": [
            ROTTERDAM,
            (50.5, 1.0),          # English Channel exit
            (36.0, -5.6),         # Gibraltar
            (33.5, 20.0),         # Eastern Med coastal leg
            (30.0, 32.6),         # Suez Canal
            (20.0, 39.0),         # mid Red Sea
            (12.6, 43.3),         # Bab-el-Mandeb
            (14.0, 55.0),         # Gulf of Aden/Arabian Sea constrained lane
            (6.0, 80.6),          # Sri Lanka gateway
            (1.2, 103.5),         # Malacca
            (22.0, 121.0),        # Luzon/Taiwan strait guidance
            (35.0, 140.0),
            YOKOHAMA
        ],
    },

    "CAPE": {
        # (a) Service-guided fallback via W-Africa lane and Mozambique Channel.
        "VAR1_service": [
            ROTTERDAM,
            (36.0, -5.6),         # Gibraltar
            (14.7, -17.5),        # Dakar / W-Africa gateway
            (-34.0, 18.5),        # Cape of Good Hope gate
            (-20.0, 40.0),        # Mozambique Channel
            (0.0, 70.0),          # Indian Ocean pivot
            (1.2, 103.5),         # Malacca
            (15.0, 120.0),        # SCS trunk
            (35.0, 140.0),
            YOKOHAMA
        ],

        # (b) Bluewater: more open-S Atlantic + open Indian Ocean legs.
        "VAR2_bluewater": [
            ROTTERDAM,
            (30.0, -20.0),        # open Atlantic pivot
            (-10.0, -25.0),       # open S-Atlantic leg
            (-34.0, 18.5),        # Cape gate
            (-30.0, 60.0),        # open S-Indian Ocean pivot
            (1.2, 103.5),         # Malacca
            (20.0, 130.0),        # open W-Pacific
            (35.0, 140.0),
            YOKOHAMA
        ],

        # (c) Coast/channel-guided: closer African coast + explicit Mozambique guidance.
        "VAR3_coast": [
            ROTTERDAM,
            (36.0, -5.6),         # Gibraltar
            (5.0, -5.0),          # Gulf of Guinea coastal lane
            (-20.0, 12.0),        # Namibia/Angola offshore lane
            (-34.0, 18.5),        # Cape gate
            (-25.0, 35.0),        # East Africa coast
            (-20.0, 40.0),        # Mozambique Channel
            (6.0, 80.6),          # Sri Lanka gateway
            (1.2, 103.5),         # Malacca
            (22.0, 121.0),
            (35.0, 140.0),
            YOKOHAMA
        ],
    },

    "NSR": {
        # (a) Service-guided: follows main Arctic “gates” + standard Pacific trunk.
        "VAR1_service": [
            ROTTERDAM,
            (71.0, 25.0),         # North Cape / Barents gateway
            (70.5, 58.0),         # Kara Gate / Novaya Zemlya
            (76.0, 102.0),        # Vilkitsky/Laptev gateway
            (72.0, 160.0),        # East Siberian Sea pivot
            (66.0, -169.0),       # Bering Strait
            (43.0, 150.0),        # NW-Pacific pivot (your fix)
            (35.0, 140.0),        # Japan gate
            YOKOHAMA
        ],

        # (b) Bluewater-oriented: keeps Arctic gates but uses more open Barents + Pacific legs.
        "VAR2_bluewater": [
            ROTTERDAM,
            (73.0, 35.0),         # open Barents pivot (north of Norway)
            (70.5, 58.0),         # Kara Gate
            (76.5, 120.0),        # more open Laptev/ESS transition
            (73.0, 150.0),        # open ESS pivot
            (66.0, -169.0),       # Bering Strait
            (43.0, 150.0),        # open Pacific pivot
            (35.0, 140.0),
            YOKOHAMA
        ],

        # (c) Coast/channel-guided: slightly more southerly, coast-hugging Arctic chain.
        "VAR3_coast": [
            ROTTERDAM,
            (70.0, 33.0),         # Kola/Barents coastal leg
            (70.5, 58.0),         # Kara Gate
            (75.0, 110.0),        # Laptev south/coastal guidance
            (71.0, 165.0),        # ESS south/coastal guidance
            (66.0, -169.0),       # Bering Strait
            (43.0, 150.0),        # NW-Pacific pivot
            (35.0, 140.0),
            YOKOHAMA
        ],
    }
}

print("Loaded WAYPOINTS with explicit philosophies:")
for c in WAYPOINTS:
    print(c)
    for v in WAYPOINTS[c]:
        key = v.split("_")[-1]
        print("  ", v, "->", PHILOSOPHY_DESC.get(key, ""))

# Flatten to scenario dict used everywhere else
SCENARIOS = {}
for corridor, vars_ in WAYPOINTS.items():
    for vname, chain in vars_.items():
        scen_key = f"{corridor}_{vname}"
        SCENARIOS[scen_key] = chain

def corridor_of(scen_key):
    return scen_key.split("_")[0]

def philosophy_of(scen_key):
    # returns SERVICE / BLUEWATER / COAST
    return scen_key.split("_")[1]
```

```python
SCENARIOS
```

```python
# ============================================================
# Variant manifest (quick reproducibility audit)
# ============================================================

manifest_rows = []
for scen_key, chain in SCENARIOS.items():
    manifest_rows.append({
        "scenario": scen_key,
        "corridor": corridor_of(scen_key),
        "philosophy": philosophy_of(scen_key),
        "n_waypoints": len(chain),
        "waypoints_chain": " → ".join([f"({lat:.1f},{lon:.1f})" for lat, lon in chain])
    })

df_manifest = pd.DataFrame(manifest_rows).sort_values(["corridor","philosophy"])
df_manifest
```

```python
df_manifest.to_csv("tables/waypoint_manifest.csv", index=False)
df_manifest.to_latex("tables/waypoint_manifest.tex", index=False)
print("Wrote tables/waypoint_manifest.csv/.tex")
```

```python
# --- 5. Core computation: GC-chain vs sea-only ---

records = []
routes_gc = {}
routes_sea = {}

for scen_key, wpts in SCENARIOS.items():
    corr = corridor_of(scen_key)

    # dateline guards for stability
    wpts_guarded = insert_dateline_guards(wpts)

    # GC-chain polyline (just between guarded macros)
    gc_poly = wpts_guarded[:]  # sequence
    gc_dist = polyline_nm(gc_poly)
    routes_gc[scen_key] = gc_poly

    records.append(dict(
        scenario=scen_key,
        corridor=corr,
        variant=scen_key.split("_")[-1],
        method="GC",
        distance_nm=gc_dist
    ))

    # SEA-only chain polyline using searoute per leg
    sea_poly_all = [wpts_guarded[0]]
    sea_dist_total = 0.0

    for a, b in zip(wpts_guarded[:-1], wpts_guarded[1:]):
        leg_poly, leg_nm = sea_leg_with_guard(a, b)
        sea_dist_total += leg_nm

        # append leg poly excluding duplicate first point
        sea_poly_all.extend(leg_poly[1:])

    routes_sea[scen_key] = sea_poly_all

    records.append(dict(
        scenario=scen_key,
        corridor=corr,
        variant=scen_key.split("_")[-1],
        method="SEA",
        distance_nm=sea_dist_total
    ))


df = pd.DataFrame(records)
df["distance_knm"] = df["distance_nm"] / 1000.0
```

```python
df
```

```python
# --- 6. Case 0 outputs: per-variant GC vs SEA + deltas ---

# pivot per scenario
piv = df.pivot_table(index=["corridor","scenario","variant"],
                     columns="method",
                     values="distance_nm").reset_index()

piv["delta_nm"]  = piv["SEA"] - piv["GC"]
piv["delta_pct"] = 100.0 * piv["delta_nm"] / piv["GC"]

# save BASELINE_all
baseline_all_csv = OUT_TABLES / "BASELINE_all.csv"
baseline_all_tex = OUT_TABLES / "BASELINE_all.tex"
piv.to_csv(baseline_all_csv, index=False)

tex_tbl = piv.copy()
tex_tbl["GC"]  = (tex_tbl["GC"]/1000).round(2)
tex_tbl["SEA"] = (tex_tbl["SEA"]/1000).round(2)
tex_tbl["delta_pct"] = tex_tbl["delta_pct"].round(1)

tex_tbl = tex_tbl.rename(columns={
    "scenario":"Variant",
    "GC":"GC (k nm)",
    "SEA":"Sea-only (k nm)",
    "delta_pct":"GC→Sea Δ (%)"
})

tex_tbl.to_latex(
    baseline_all_tex,
    index=False,
    float_format="%.2f",
    escape=False
)

print("Wrote:", baseline_all_csv, baseline_all_tex)
```

```python
# --- 7. Case 1 outputs: sea-only medians + within-corridor spread ---

sea_only = df[df.method=="SEA"].copy()

med = sea_only.groupby("corridor")["distance_nm"].agg(["median","min","max","std"]).reset_index()
med["cv_pct"] = 100.0 * med["std"]/med["median"]

med_csv = OUT_TABLES / "BASELINE_corridor_medians.csv"
med_tex = OUT_TABLES / "BASELINE_corridor_medians.tex"
med.to_csv(med_csv, index=False)

med_tbl = med.copy()
med_tbl["median"] = (med_tbl["median"]/1000).round(2)
med_tbl["min"]    = (med_tbl["min"]/1000).round(2)
med_tbl["max"]    = (med_tbl["max"]/1000).round(2)
med_tbl["cv_pct"] = med_tbl["cv_pct"].round(1)

med_tbl = med_tbl.rename(columns={
    "median":"Median (k nm)",
    "min":"Min (k nm)",
    "max":"Max (k nm)",
    "cv_pct":"CV (%)"
})

med_tbl.to_latex(
    med_tex,
    index=False,
    float_format="%.2f",
    escape=False
)

print("Wrote:", med_csv, med_tex)
```

```python
# --- 8. Figures ---

# 8.1 GC distances bar plot (Case 0)
gc_only = df[df.method=="GC"].copy()
order = ["NSR","SUEZ","CAPE"]  # shortest to longest narrative

fig, ax = plt.subplots(figsize=(8,4.5))
for corr in order:
    sub = gc_only[gc_only.corridor==corr].sort_values("variant")
    ax.bar(
        [f"{corr}_{v}" for v in sub.variant],
        sub.distance_nm/1000.0,
        label=corr
    )

ax.set_ylabel("Distance (k nm)")
ax.set_title("Geometric (GC-chain) distance by corridor/variant")
ax.grid(axis="y", alpha=0.3)
ax.legend()
plt.xticks(rotation=30, ha="right")
plt.tight_layout()

gc_fig = OUT_FIGS / "case01_distance_bars_gc.png"
plt.savefig(gc_fig, dpi=220)
plt.show()

# 8.2 Sea-only distances bar plot (Case 1)
fig, ax = plt.subplots(figsize=(8,4.5))
for corr in order:
    sub = sea_only[sea_only.corridor==corr].sort_values("variant")
    ax.bar(
        [f"{corr}_{v}" for v in sub.variant],
        sub.distance_nm/1000.0,
        label=corr
    )

ax.set_ylabel("Distance (k nm)")
ax.set_title("Sea-only distance by corridor/variant")
ax.grid(axis="y", alpha=0.3)
ax.legend()
plt.xticks(rotation=30, ha="right")
plt.tight_layout()

sea_fig = OUT_FIGS / "BASELINE_distance.png"
plt.savefig(sea_fig, dpi=220)
plt.show()
```

```python
# 8.3 Dumbbell/slopegraph: GC vs SEA by variant (routing-method effect)
fig, ax = plt.subplots(figsize=(8,5))

piv_plot = piv.sort_values(["corridor","variant"])
ypos = np.arange(len(piv_plot))

ax.hlines(ypos, piv_plot["GC"]/1000, piv_plot["SEA"]/1000, color="k", alpha=0.25)
ax.plot(piv_plot["GC"]/1000, ypos, "o", label="GC")
ax.plot(piv_plot["SEA"]/1000, ypos, "o", label="Sea-only")

ax.set_yticks(ypos)
ax.set_yticklabels(piv_plot["scenario"])
ax.invert_yaxis()
ax.set_xlabel("Distance (k nm)")
ax.set_title("Routing-method effect: GC-chain vs sea-only")
ax.grid(axis="x", alpha=0.3)
ax.legend()
plt.tight_layout()

delta_fig = OUT_FIGS / "BASELINE_gc_vs_sea_dumbbell.png"
plt.savefig(delta_fig, dpi=220)
plt.show()
```

```python
# --- 9. Write warning log ---
warn_path = OUT_LOGS / "warnings_case01.txt"
warn_path.write_text("\n".join(WARNLOG))
print("Warnings saved to:", warn_path)

print("\nDone. Produced:")
print("-", baseline_all_csv)
print("-", baseline_all_tex)
print("-", med_csv)
print("-", med_tex)
print("-", gc_fig)
print("-", sea_fig)
print("-", delta_fig)
print("-", warn_path)
```

```python
# ============================================================
# B) Heatmap for GC->SEA delta (%) by corridor/variant
# ============================================================

# We already computed piv with delta_pct.
heat = piv.copy()
heat["variant_label"] = heat["corridor"] + "_" + heat["variant"]

# corridor x variant matrix of delta_pct
mat = heat.pivot(index="corridor", columns="variant", values="delta_pct").loc[["NSR","SUEZ","CAPE"]]

fig, ax = plt.subplots(figsize=(6.5,3.5))
im = ax.imshow(mat.values, aspect="auto")

ax.set_xticks(np.arange(mat.shape[1]))
ax.set_xticklabels(mat.columns)

ax.set_yticks(np.arange(mat.shape[0]))
ax.set_yticklabels(mat.index)

# annotate cells
for i in range(mat.shape[0]):
    for j in range(mat.shape[1]):
        ax.text(j, i, f"{mat.values[i,j]:.1f}%", ha="center", va="center")

ax.set_title("GC-chain → Sea-only distance delta (%)")
plt.colorbar(im, ax=ax, label="Δ (%)")
plt.tight_layout()

heat_path = OUT_FIGS / "BASELINE_delta_heatmap_pct.png"
plt.savefig(heat_path, dpi=220)
plt.show()

print("Wrote:", heat_path)
```
